# **Features That Can Act As TAGS**

```python
#genres
#id-to make website
#keywords
#title
#Overview
#cast
#crew

movies=movies[['genres','id','keywords','title','overview','cast','crew']]
```

```python
movies.dropna(inplace=True)
```

```python
import ast
```

```python
def convert(list):
    L=[]
    for i in ast.literal_eval(list):
        L.append(i['name'])
    return L
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies['keywords']=movies['keywords'].apply(convert)
```

```python
def convert_(list):
    L=[]
    counter=0
    for i in ast.literal_eval(list):
        L.append(i['name'])
        counter=counter+1
        if counter==3:
            break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert_)
```

```python
def convert1(list):
    L=[]
    for i in ast.literal_eval(list):
        if i['job']=='Director':
            L.append(i['name'])
            break
    return L
```

```python
movies['crew']=movies['crew'].apply(convert1)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies['tags']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
final_df=movies[['id','title','tags']]
```

```python
final_df
```

```python
final_df['tags']=final_df['tags'].apply(lambda x:" ".join(x))
```

```python
final_df['tags']=final_df['tags'].apply(lambda x:x.lower())
```

```python
import nltk 
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(object):
    l=[]
    for i in object.split():
        l.append(ps.stem(i))

    return " ".join(l)
        
```

```python
final_df['tags']=final_df['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vec=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors=vec.fit_transform(final_df['tags']).toarray()
```

```python
vec.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise  import cosine_similarity
scores = cosine_similarity(vectors)
```
